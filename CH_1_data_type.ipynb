```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("a == b :", a == b)
print("a != b :", a != b)
print("a > b  :", a > b)
print("a < b  :", a < b)
print("a >= b :", a >= b)
print("a <= b :", a <= b)
```

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("AND:", a > 0 and b > 0)
print("OR :", a > 0 or b > 0)
print("NOT:", not(a > 0))
```
