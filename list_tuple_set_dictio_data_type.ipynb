# **tuple**

Key Features:
Ordered (maintains insertion order)
 Allows duplicate elements

 Can contain mixed data types

 Immutable (cannot be modified after creation)

 No methods to add/remove elements

```python
#creating a tuple
a = ('python', 'program',1,2, 3.3, True)
#another approach
b = 'python' , 'program',1
print(a)
print(type(b))
```

```python
# Basic tuple
coordinates = (10, 20)
person = ("John", 25, "Engineer")

# Single element tuple (note the comma)
single = (42,)  # This is a tuple
not_tuple = (42)  # This is just an integer

print(type(single))    # <class 'tuple'>
print(type(not_tuple)) # <class 'int'>
```

```python
# Method 1: Parentheses
tuple1 = (1, 2, 3)

# Method 2: Without parentheses (tuple packing)
tuple2 = 1, 2, 3  # (1, 2, 3)

# Method 3: tuple() constructor
tuple3 = tuple([1, 2, 3])  # (1, 2, 3)
tuple4 = tuple("hello")     # ('h', 'e', 'l', 'l', 'o')

# Method 4: From other iterables
tuple5 = tuple(range(3))  # (0, 1, 2)
```

```python
tuple1 = (1, 2)
tuple2 = (3, 4)
combined = tuple1 + tuple2  # (1, 2, 3, 4)
print(combined)
```

```python
my_tuple2 = (1, 2, 3, 'programmimg_hub') #access elements
for x in my_tuple2:
    print(x)
print(my_tuple2)
print(my_tuple2[0])
print(my_tuple2[:])
print(my_tuple2[3][4])
```

```python
numbers = (1, 2, 2, 3, 2, 4)
print(numbers.count(2))  # 3
print(numbers.count(9))  # 0

# With strings
words = ("apple", "banana", "apple", "cherry")
print(words.count("apple"))  # 2
```

```python
data = (1,10, "hello", 30, 20, 40)

print(data.index("hello"))       # 2
print(data.index(20, 2))    # 3 (search from index 2)
print(data.index(100))       # 2

# data.index(99) → ValueError: tuple.index(x): x not in tuple
```

# **Set**
Key Characteristics:

Unordered: Elements have no specific order

Unique: No duplicate elements allowed

Mutable: You can add/remove elements

Elements must be hashable (immutable)

```python
my_set = {1, 2, 3, 4, 5, 5, 5} #create set
print(my_set)
my_set = {1, 2, 3}
my_set.add((1, 2)) #add element to set
print(my_set)
```

```python
# Creating a set
my_set = {1, 2, 3, 4, 5}
print("Initial Set:", my_set)

# Adding elements to a set
my_set.add(6)
print("Set after adding 6:", my_set)

# Trying to add a duplicate element
my_set.add(6)  # Won't be added as it's already in the set
print("Set after trying to add 6 again:", my_set)

# Removing an element from the set
my_set.remove(3)
print("Set after removing 3:", my_set)
```

```python
# Checking membership
print("Is 2 in the set?", 2 in my_set)
print("Is 7 in the set?", 7 in my_set)


# Set operations
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
set3 = {1, 2, 3, 4, 11, 12}


# Union
union_set = set2.union(set1,set3)
print("Union of set1 and set2:", union_set)

# Intersection
intersection_set = set2.intersection(set1,set3)
print("Intersection of set1 and set2:", intersection_set)
```

```python
# Difference
difference_set = set1.difference(set2,set3)
print("Difference of set1 and set2:", difference_set)

# Clearing the set
my_set.clear()
print("Cleared Set:", my_set)
```

```python
# Creating sets
empty_set = set()
numbers = {1, 2, 3, 4, 5}
fruits = {'apple', 'banana', 'orange'}
mixed = {1, 'hello', 3.14, (1, 2)}  # All elements must be hashable

print("Original set:", numbers)

# Basic operations
numbers.add(6)
print("After adding 6:", numbers)

numbers.remove(3)
print("After removing 3:", numbers)

# Set operations
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

print("Union:", set_a | set_b)          # {1, 2, 3, 4, 5, 6, 7, 8}
print("Intersection:", set_a & set_b)   # {4, 5}
print("Difference:", set_a - set_b)     # {1, 2, 3}
print("Symmetric Difference:", set_a ^ set_b)  # {1, 2, 3, 6, 7, 8}

# Membership testing
print("Is 2 in set_a?", 2 in set_a)  # True
print("Is 9 in set_a?", 9 in set_a)  # False

# Set comprehensions
squares = {x**2 for x in range(10)}
print("Squares set:", squares)
```

#**dictionary**
Key Characteristics:

Unordered (Python 3.7+ maintains insertion order)

Keys must be unique and hashable

Values can be any Python object

Mutable: You can add/remove key-value pairs

```python
# Creating a dictionary
# Basic dictionary
person = {
    "name": "John",
    "age": "john",
    "city": "New York"
}

# Empty dictionary
empty_dict = {}
print(person)
```

```python
# Method 1: Curly braces
dict1 = {"name": "Alice", "age": 25}

# Method 2: dict() constructor
dict2 = dict(name=["Bob", "alice"], age=[30,18])
print(type(dict2))

# Method 3: From list of tuples
dict3 = dict([("name", "Charlie"), ("age", 35)])

# Method 4: Dictionary comprehension
dict4 = {x: x**2 for x in range(5)}  # {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

```python
person = {"name": "John", "age": 30}

# Using key
print(person["name"])  # John

# Using get() (safer)
print(person.get("age"))  # 30
print(person.get("city", "key is not found"))  # Not found (default value)

# person["city"]
```

```python
# Creating dictionaries
empty_dict = {}
person = {'name': 'John', 'age': 30, 'city': 'New York'}
numbers = dict(one=1, two=2, three=3)

print("Person dictionary:", person)

# Accessing values
print("Name:", person['name'])
print("Age:", person.get('age'))
print("Salary:", person.get('salary', 'Not specified'))  # Default value

# Adding/updating
person['email'] = 'john@example.com'
person['age'] = 31
print("Updated person:", person)

# Dictionary methods
print("Keys:", list(person.keys()))
print("Values:", list(person.values()))
print("Items:", list(person.items()))

# Dictionary comprehensions
squares = {x: x**2 for x in range(1, 6)}
print("Squares dictionary:", squares)

# Nested dictionaries
students = {
    'student1': {'name': 'Alice', 'grades': [85, 90, 78]},
    'student2': {'name': 'Bob', 'grades': [92, 88, 95]}
}

print("Alice's grades:", students['student1']['grades'])

# Merging dictionaries (Python 3.9+)
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}
merged = dict1 | dict2
print("Merged dictionary:", merged)
```

```python
# Create list of tuples
student_list = [
    (101, 'Alice', 'Software Developer'),
    (102, 'Bob', 'Data Analyst'),
    (103, 'Charlie', 'Project Manager')
]

# Convert to dictionary using comprehension
student_dict = {
    f'stu{i+1}': {'stID': stu[0], 'FName': stu[1], 'job': stu[2]}
    for i, stu in enumerate(student_list)
}

# Print formatted output
print("Employee Dictionary:")
for stu_key, stu_info in student_dict.items():
    print(f"{stu_key}: {stu_info}")
```
