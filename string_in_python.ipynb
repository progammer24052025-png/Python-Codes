```python
text = "Python Programming"
#(start:stop(n-1):step)
print(text[0:5:2])    # 'Python' (from index 0 to 5)
print(text[9:])     # 'Programming' (from index 7 to end)
print(text[:6])     # 'Python' (from start to index 5)
print(text[::2])    # 'Pto rgamn' (every 2nd character)
print(text[7:3:-1])   # 'gnimmargorP nohtyP' (reverse string)
```

```python
text = "!!!Hello, World!!!"

# strip() removes from both sides
trimmed = text.strip('!')
print(trimmed)  # Output: "Hello, World!"

# lstrip() removes only from the left
left_trimmed = text.lstrip()
print(left_trimmed)  # Output: "Hello, World!   "

# rstrip() removes only from the right
right_trimmed = text.rstrip()
print(right_trimmed)  # Output: "   Hello, World!"

text = "@##@#(Hello###!!~"
print(text.strip('@#(~!'))  # Output: "Hello"
```

```python
text = "apple,banana,cherry"
fruits = text.split(",")
print(fruits)  # Output: ['apple', 'banana', 'cherry']

# Splitting with maxsplit (limits the number of splits)
text = "one two three four FIVE"
words = text.split(" ", 3)
print(words)  # Output: ['one', 'two', 'three four']
```

```python
text = "one two three four"
words = text.split(" ", 2)
print(words)  # Output: ['one,two', 'three', 'four']
```

```python
words = ("Hello", "World", "Python")
joined = " ".join(words)
print(joined)  # Output: "Hello World Python"

# Joining with a different separator
joined_with_comma = ", ".join(words)
print(joined_with_comma)  # Output: "Hello, World, Python"

# Joining characters of a string
chars = ["1", "2", "ABC"]
joined_chars = "-".join(chars)
print(joined_chars)  # Output: "a-b-c"
```

```python
text = "Python PrograPring"

# find() returns -1 if not found
print(text.find("Pr",7, 9))  # Output: 7
print(text.find("Java"))  # Output: -1

# index() raises an error if not found
print(text.index("Pro"))  # Output: 7
print(text.index("Java"))
```

```python
text = "Hello, World!"

# Basic replacement
new_text = text.replace("Hello", "Python")
print(new_text)

# Limiting replacements
text = "ha ha ha"
new_text = text.replace("ha", "ho", 2)
print(new_text)  # Output: "ho ho ha"
```
